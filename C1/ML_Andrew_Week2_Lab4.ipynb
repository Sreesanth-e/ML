```python
import sys
sys.path.append('../data')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from lab_utils_multi import zscore_normalize_features, run_gradient_descent_feng
np.set_printoptions(precision=2)
```

```python
x = np.arange(0,20,1)
y = 1 + x**2
X = x.reshape(-1,1)
print(y)
model_w, model_b = run_gradient_descent_feng(X,y,iterations=1000 , alpha=1e-2)

plt.scatter(x,y,marker='x',c = 'r' , label = "Actual Value"); plt.title("No Engineering")
plt.plot(x,X@model_w + model_b , label = "Predicted Value"); plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.show()
```

```python
x = np.arange(0,20,1)
y = 1 + x**3
X = x**3
X = X.reshape(-1,1)
print(y)
model_w, model_b = run_gradient_descent_feng(X,y,iterations=1000 , alpha=1e-7)

plt.scatter(x,y,marker='x',c = 'r' , label = "Actual Value"); plt.title("Feature Engineering")
plt.plot(x,X@model_w + model_b , label = "Predicted Value"); plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.show()
```

```python
x = np.arange(0,20,1)
y = x**2
X = np.c_[x,x**2,x**3]

X_features = ['x' , 'x^2' , 'x^3']
```

```python
fig, ax = plt.subplots(1,3 , figsize = (18,5) , sharey= True)
for i in range(len(ax)):
    ax[i].scatter(X[:,i],y)
    ax[i].set_xlabel(X_features[i])
ax[0].set_ylabel("y")
plt.show()
```

```python
x = np.arange(0,20,1)
X = np.c_[x,x**2,x**3]
print("Peak  to peak  range in column in raw    X : " , np.ptp(X,axis=0))
X = zscore_normalize_features(X)
print("Peak  to peak  range in column in normalized    X : " , np.ptp(X,axis=0))
```

```python

model_w , model_b =run_gradient_descent_feng(X,y, iterations= 100000, alpha= 5e-1)
plt.scatter(x,y , c = 'b' , label = "Actual value")
plt.title("Normalized x , x^2 , x^3")
plt.plot(x,X@model_w + model_b , c= 'r', label = "Prediction")
plt.legend()
plt.show()
```
