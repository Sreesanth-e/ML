```python
def compute_linear_reg(X,y,w,b ,lambda_ = 1):
    m = X.shape[0]
    n = len(w)
    cost = 0.
    regu = 0.
    for i in range(m):
        f_wb_i = np.dot(X[i] , w) + b
        cost = cost + (f_wb_i - y[i])**2
    cost = cost / (2 * m)

    for j in range(n):
        regu= regu + w[j] ** 2
    regu_fun  = (lambda_/ (2 * m)) * regu

    total_cost = cost + regu_fun

    return total_cost
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0.,1.,0.,1.,0.])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,) - 0.5
b_tmp = 0.5
lamba_tmp = 0.7
cost_tmp = compute_linear_reg(X_tmp,y_tmp,w_tmp,b_tmp,lamba_tmp)
print(f"Regularized Cost : {cost_tmp}")
```

```python
def sigmoid(z):
    g = 1.0/(1.0+np.exp(-z))
    return g
```

```python
def compute_logistic_reg(X,y,w,b,lambda_ = 1):
    m,n = X.shape


    cost = 0.
    regu = 0.

    for i in range(m):
        z = np.dot(X[i] , w) + b
        f_wb_i = sigmoid(z)

        cost += -y[i]* np.log(f_wb_i) - (1 - y[i])* np.log(1 - f_wb_i)
    cost = cost / m

    for j in range(n):
        regu = regu + (w[j] ** 2)
    regu_fun = (lambda_/(2 * m)) * regu

    total_cost = cost + regu_fun

    return total_cost
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0.,1.,0.,1.,0.])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,) - 0.5
b_tmp = 0.5
lamba_tmp = 0.7
cost_tmp = compute_logistic_reg(X_tmp,y_tmp,w_tmp,b_tmp,lamba_tmp)
print(f"Regularized Cost : {cost_tmp}")
```

```python
def compute_gradient_linear_regu(X,y,w,b,lambda_):
    m,n = X.shape

    dj_dw = np.zeros((n))
    dj_db = 0

    for i in range(m):
        z = np.dot(X[i], w) + b
        err = z - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + err * X[i,j]
        dj_db = dj_db + err
    dj_dw = dj_dw/m
    dj_db = dj_db/m
    for j in range(n):
        dj_dw[j] = dj_dw[j] + (lambda_/m) * w[j]

    return dj_dw,dj_db
```

```python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0.,1.,0.,1.,0.])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lamba_tmp = 0.7
dj_dw_tmp,dj_db_tmp = compute_gradient_linear_regu(X_tmp,y_tmp,w_tmp,b_tmp,lamba_tmp)
print(f"Regularized dj_db : {dj_db_tmp}")
print(f"Regularized dj_dw : {dj_dw_tmp.tolist()}")
```

```python
def compute_gradient_logistic_regu(X,y,w,b,lambda_):
    m,n = X.shape

    dj_dw = np.zeros((n,))
    dj_db = 0

    for i in range(m):
        z = np.dot(X[i], w) + b
        err = sigmoid(z)-y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + err * X[i,j]
        dj_db = dj_db + err
    dj_dw = dj_dw/m
    dj_db = dj_db/m
    for j in range(n):
        dj_dw[j] = dj_dw[j] + (lambda_/m) * w[j]

    return dj_dw,dj_db
```
