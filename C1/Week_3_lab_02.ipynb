```python
def sigmoid(z):
    g = 1/(1 + np.exp(-z))
    return g
```

```python
z_tmp = np.arange(-10,11)
y = sigmoid(z_tmp)
np.set_printoptions( precision= 3)
print("Input(z) , Output(Sigmoid)")
print(np.c_[z_tmp,y])
```

```python
fig , ax = plt.subplots(1,1,figsize= (5,4))

ax.plot(z_tmp,y)
ax.set_title("Sigmoid Function")
ax.set_ylabel("Sigmoid(z)")
ax.set_xlabel("Input values")
draw_vthresh(ax,0)
```
