```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from lab_utils_multi import zscore_normalize_features, run_gradient_descent_feng
np.set_printoptions(precision=2)
```

```python
x = np.arange(0,20,1)
y = 1 + x**2
X = x.reshape(-1,1)
print(y)
model_w, model_b = run_gradient_descent_feng(X,y,iterations=1000 , alpha=1e-2)

plt.scatter(x,y,marker='x',c = 'r' , label = "Actual Value"); plt.title("No Engineering")
plt.plot(x,X@model_w + model_b , label = "Predicted Value"); plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.show()
```

```python
x = np.arange(0,20,1)
y = 1 + x**3
X = x**3
X = X.reshape(-1,1)
print(y)
model_w, model_b = run_gradient_descent_feng(X,y,iterations=1000 , alpha=1e-7)

plt.scatter(x,y,marker='x',c = 'r' , label = "Actual Value"); plt.title("Feature Engineering")
plt.plot(x,X@model_w + model_b , label = "Predicted Value"); plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.show()
```
