```python
w = 200
b = 100
print(f"weights : {w}")
print(f"bias : {b}")
```

```python
def compute_model_output(x , w, b):
    m  = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b

    return f_wb
```

```python
f_wb = compute_model_output(x_array,w,b)
plt.plot(x_array,f_wb , c = 'g' , label = "Our prediction")
plt.scatter(x_array,y_array, c = 'b' , label = "Actual labels" , marker='x')
plt.title("Housing price")
plt.ylabel("1000 s of dollars")
plt.xlabel("Size in sqft")
plt.legend()
plt.show()
```

```python
def compute_cost(x ,y, w, b):
    m  = x.shape[0]
    f_wb = np.zeros(m)
    cost_sum = 0
    for i in range(m):
        f_wb[i] = w * x[i] + b
        cost  = (f_wb[i] - y[i]) ** 2
        cost_sum = cost_sum + cost
        total_cost = (1/(2 * m)) * cost_sum

    return total_cost
```

```python
print(compute_cost(x_array,y_array,w,b))
```
