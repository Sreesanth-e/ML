```python
a = np.arange(1,10)
print(f"a vector(1D) : {a}")
print("Slicing : ")
c = a[0:5:1]
print("a[0:5:1] : ",c)
c = a[2:7:2]
print("a[2:7:2] : ",c)
c = a[3:]
print("3: : ",c)
c = a[:3]
print(":3 : ",c)
c = a[:]
print(": : ",c)
```

## Vector dot Product

```python
def My_dot(a,b):
    x = 0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
a = np.array([1,2,3,4])
b = np.array([-1,4,4,2])
print(f"My_dot(a,b) : {My_dot(a,b)}")
```

```python
## Dot Product
a = np.array([1,2,3,4])
b = np.array([-1,4,4,2])
c = np.dot(a,b)
print(f"Using Dot product : {c}")
c = np.dot(b,a)
print(f"Using Dot product : {c}")
```

```python
a = np.random.rand(10000000)
b = np.random.rand(10000000)
tic = time.time()
print(f"Without Vectorization : {My_dot(a,b)}")
toc = time.time()
print(f"Without Vectorization (Time): {1000 *(toc - tic):.4f} ms")
tic = time.time()
print(f"With Vectorization : {np.dot(a,b)}")
toc = time.time()
print(f"With Vectorization (Time): {1000 *(toc - tic):.4f} ms")
```
