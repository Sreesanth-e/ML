```python
def zscore_normalization(X):
    mu = np.mean(X, axis= 0)
    sigma = np.std(X, axis=0)
    X_norm = (X - mu) / sigma
    return (X_norm,mu,sigma)


```

```python
mu = np.mean(X_train,axis=0)
sigma = np.std(X_train, axis=0)
Zscore_mean = (X_train - mu)
Zscore = (X_train - mu) / sigma
```

```python
fig, ax = plt.subplots(1,3, figsize = (12,3))
ax[0].scatter(X_train[:,0], X_train[:,3])
ax[0].set_xlabel(X_featuers[0])
ax[0].set_ylabel(X_featuers[3])
ax[0].set_title("unnormalized")
ax[0].axis('equal')
ax[1].scatter(Zscore_mean[:,0], Zscore_mean[:,3])
ax[1].set_xlabel(X_featuers[0])
ax[0].set_ylabel(X_featuers[3])
ax[1].set_title(r"X - $\mu$")
ax[1].axis('equal')
ax[2].scatter(Zscore[:,0], Zscore[:,3])
ax[2].set_xlabel(X_featuers[0])
ax[0].set_ylabel(X_featuers[3])
ax[2].set_title(r"Z - score normalized")
ax[2].axis('equal')
plt.tight_layout(rect=[0,0.03,1,0.95])
plt.show()
```

```python
X_norm,X_mu,X_sigma = zscore_normalization(X_train)
print(f"X_mu ; {X_mu} , X_sigma : {X_sigma}")
print(f"raw  : X: {np.ptp(X_train,axis=0)}")
print(f"Normalized  : X: {np.ptp(X_norm,axis=0)}")
```

```python
fig , ax  =plt.subplots(1,4, figsize = (12,2))
for i in range(len(ax)):
    norm_plot(ax[i], X_train[:,i])
    ax[i].set_xlabel(X_featuers[i])
ax[0].set_ylabel("Count")
fig.suptitle("Distribution of featues before Normalization : ")
fig , ax  =plt.subplots(1,4, figsize = (12,2))
for i in range(len(ax)):
    norm_plot(ax[i], X_norm[:,i])
    ax[i].set_xlabel(X_featuers[i])
ax[0].set_ylabel("Count")
fig.suptitle("Distribution of featues after Normalization : ")
```

```python
w_norm , b_norm , hist = run_gradient_descent(X_norm,Y_train,1000,1.0e-1)
```

```python
m = X_norm.shape[0]
yp = np.zeros(m)
```

```python
for i in range(m):
    yp[i] = np.dot(X_norm[i] , w_norm) + b_norm

fig , ax = plt.subplots(1,4 , figsize = (12,3) , sharey= True)
for i in range(len(ax)):
    ax[i].scatter(X_train[:,i] , Y_train,label = "target")
    ax[i].set_ylabel(X_featuers[i])
    ax[i].scatter(X_train[:,i], yp, color =   dlc["dlorange"] , label = "Predict")
ax[0].set_ylabel("Price"); ax[0].legend()
plt.show()
```
